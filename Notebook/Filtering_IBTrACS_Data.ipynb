```python
# Select the period we want for the ML project
model_track = track[
    (track["SEASON"] >= 2000) &
    (track["SEASON"] <= 2025)
].copy()

print("========== MODEL DATA SUMMARY ==========")

print("Observations:", len(model_track))
print("Unique cyclones:", model_track["SID"].nunique())

print(
    "Years:",
    model_track["SEASON"].min(),
    "to",
    model_track["SEASON"].max()
)

print(
    "Latitude:",
    model_track["LAT"].min(),
    "to",
    model_track["LAT"].max()
)

print(
    "Longitude:",
    model_track["LON"].min(),
    "to",
    model_track["LON"].max()
)

print("\nMissing values:")
print(model_track[[
    "LAT", "LON", "USA_WIND", "USA_PRES"
]].isna().sum())

print("\nCyclones per year:")
print(
    model_track[
        ["SID", "SEASON", "NAME"]
    ]
    .drop_duplicates()
    .groupby("SEASON")
    .size()
    .to_string()
)
```

```python
# Define the North Indian Ocean model region

NORTH = 30.0
SOUTH = 0.0
WEST = 40.0
EAST = 110.0

noi_track = model_track[
    (model_track["LAT"] >= SOUTH) &
    (model_track["LAT"] <= NORTH) &
    (model_track["LON"] >= WEST) &
    (model_track["LON"] <= EAST)
].copy()

print("========== NORTH INDIAN OCEAN FILTER ==========")

print("Observations:", len(noi_track))
print("Unique cyclones:", noi_track["SID"].nunique())

print(
    "Latitude:",
    noi_track["LAT"].min(),
    "to",
    noi_track["LAT"].max()
)

print(
    "Longitude:",
    noi_track["LON"].min(),
    "to",
    noi_track["LON"].max()
)

print(
    "Years:",
    noi_track["SEASON"].min(),
    "to",
    noi_track["SEASON"].max()
)

print("\nMissing values:")
print(
    noi_track[
        ["LAT", "LON", "USA_WIND", "USA_PRES"]
    ].isna().sum()
)
```

```python
noi_storms = (
    noi_track[
        ["SID", "SEASON", "NAME"]
    ]
    .drop_duplicates()
    .sort_values(["SEASON", "NAME"])
)

print("Number of storms:", len(noi_storms))
print()
print(noi_storms.to_string(index=False))
```

```python
# Final North Indian Ocean region
FINAL_NORTH = 30.0
FINAL_SOUTH = 0.0
FINAL_WEST = 40.0
FINAL_EAST = 100.0

final_track = model_track[
    (model_track["LAT"] >= FINAL_SOUTH) &
    (model_track["LAT"] <= FINAL_NORTH) &
    (model_track["LON"] >= FINAL_WEST) &
    (model_track["LON"] <= FINAL_EAST)
].copy()

final_storms = (
    final_track[["SID", "SEASON", "NAME"]]
    .drop_duplicates()
)

print("========== FINAL MODEL REGION ==========")
print("Observations:", len(final_track))
print("Unique cyclones:", final_track["SID"].nunique())
print("Years:", final_track["SEASON"].min(), "to", final_track["SEASON"].max())

print("\nCoordinates:")
print("Latitude :", final_track["LAT"].min(), "to", final_track["LAT"].max())
print("Longitude:", final_track["LON"].min(), "to", final_track["LON"].max())

print("\nMissing labels:")
print(final_track[["USA_WIND", "USA_PRES"]].isna().sum())

print("\nStorms by year:")
print(
    final_storms.groupby("SEASON")
    .size()
    .to_string()
)
```

```python
# Create one row per cyclone with its start/end time and spatial extent

storm_ranges = (
    final_track
    .groupby(["SID", "SEASON", "NAME"])
    .agg(
        start_time=("ISO_TIME", "min"),
        end_time=("ISO_TIME", "max"),
        min_lat=("LAT", "min"),
        max_lat=("LAT", "max"),
        min_lon=("LON", "min"),
        max_lon=("LON", "max"),
        observations=("ISO_TIME", "count")
    )
    .reset_index()
)

print("Number of cyclones:", len(storm_ranges))
print("\nFirst 20 cyclones:")
print(storm_ranges.head(20).to_string(index=False))
```

```python
# ============================================
# ERA5 DATA REQUIREMENT ANALYSIS
# ============================================

# Work only with observations that have valid time/location
era5_needed = final_track.dropna(
    subset=["ISO_TIME", "LAT", "LON"]
).copy()

# ERA5 is hourly, so round each observation to its hour
era5_needed["ERA5_TIME"] = era5_needed["ISO_TIME"].dt.floor("h")

# Number of unique time points
unique_times = era5_needed["ERA5_TIME"].nunique()

# Number of unique cyclone/time combinations
unique_storm_times = era5_needed[
    ["SID", "ERA5_TIME"]
].drop_duplicates().shape[0]

print("========== ERA5 REQUIREMENT ==========")

print("IBTrACS observations:", len(era5_needed))
print("Unique ERA5 timestamps:", unique_times)
print("Unique cyclone-time pairs:", unique_storm_times)

print("\nERA5 time range:")
print(era5_needed["ERA5_TIME"].min())
print("to")
print(era5_needed["ERA5_TIME"].max())

print("\nSpatial range:")
print("Latitude :", era5_needed["LAT"].min(), "to", era5_needed["LAT"].max())
print("Longitude:", era5_needed["LON"].min(), "to", era5_needed["LON"].max())

# Count observations per year
year_counts = (
    era5_needed
    .groupby(era5_needed["ERA5_TIME"].dt.year)
    .size()
)

print("\nObservations requiring ERA5 by year:")
print(year_counts.to_string())
```

```python
# ============================================
# APPROXIMATE ERA5 GRID POINTS
# ============================================

# ERA5 atmospheric grid spacing = 0.25 degrees

era5_needed["ERA5_LAT"] = (
    np.round(era5_needed["LAT"] / 0.25) * 0.25
)

era5_needed["ERA5_LON"] = (
    np.round(era5_needed["LON"] / 0.25) * 0.25
)

unique_grid_points = era5_needed[
    ["ERA5_LAT", "ERA5_LON"]
].drop_duplicates()

print("========== ERA5 GRID ==========")

print(
    "Unique approximate ERA5 grid points:",
    len(unique_grid_points)
)

print(
    "Latitude grid range:",
    unique_grid_points["ERA5_LAT"].min(),
    "to",
    unique_grid_points["ERA5_LAT"].max()
)

print(
    "Longitude grid range:",
    unique_grid_points["ERA5_LON"].min(),
    "to",
    unique_grid_points["ERA5_LON"].max()
)
```

```python
# ============================================
# CREATE ERA5 LOOKUP TABLE
# ============================================

era5_lookup = final_track[
    ["SID", "SEASON", "NAME", "ISO_TIME",
     "LAT", "LON", "USA_WIND", "USA_PRES"]
].copy()

# ERA5 is on a 0.25° grid.
# This is only a preliminary grid estimate.
era5_lookup["ERA5_LAT"] = (
    np.round(era5_lookup["LAT"] / 0.25) * 0.25
)

era5_lookup["ERA5_LON"] = (
    np.round(era5_lookup["LON"] / 0.25) * 0.25
)

# ERA5 time corresponding to each IBTrACS observation
era5_lookup["ERA5_TIME"] = (
    era5_lookup["ISO_TIME"].dt.floor("h")
)

print("ERA5 lookup table created!")
print("Rows:", len(era5_lookup))

print("\nFirst 10 rows:")
print(
    era5_lookup.head(10).to_string(index=False)
)
```

```python
# ============================================
# UNIQUE ERA5 REQUEST COMBINATIONS
# ============================================

unique_requests = (
    era5_lookup[
        ["ERA5_TIME", "ERA5_LAT", "ERA5_LON"]
    ]
    .drop_duplicates()
    .sort_values(
        ["ERA5_TIME", "ERA5_LAT", "ERA5_LON"]
    )
    .reset_index(drop=True)
)

print("========================================")
print("UNIQUE ERA5 TIME + LOCATION COMBINATIONS")
print("========================================")

print("Total unique combinations:",
      len(unique_requests))

print("\nFirst 20:")
print(
    unique_requests.head(20).to_string(index=False)
)
```

```python
# ============================================================
# FULL IBTrACS + ERA5 EXTRACTION
# ============================================================

import time
from IPython.display import clear_output

# Variables we need
ERA5_VARS = ["u10", "v10", "t2m", "d2m", "msl", "sst"]

# Padding around each cyclone's bounding box
PADDING = 0.5

# Store results here
all_results = []

total_storms = len(storm_ranges)

print(f"Starting ERA5 extraction for {total_storms} cyclones...")
print("=" * 60)

for storm_num, (_, storm) in enumerate(storm_ranges.iterrows(), start=1):

    sid = storm["SID"]

    # --------------------------------------------------------
    # Get all IBTrACS observations belonging to this cyclone
    # --------------------------------------------------------

    storm_obs = era5_lookup[
        era5_lookup["SID"] == sid
    ].copy()

    if len(storm_obs) == 0:
        continue

    # --------------------------------------------------------
    # Define padded ERA5 extraction region
    # --------------------------------------------------------

    lat_min = max(0, storm["min_lat"] - PADDING)
    lat_max = min(30, storm["max_lat"] + PADDING)

    lon_min = max(40, storm["min_lon"] - PADDING)
    lon_max = min(100, storm["max_lon"] + PADDING)

    # --------------------------------------------------------
    # Retrieve only this cyclone's time + spatial block
    # --------------------------------------------------------

    try:

        storm_era5 = ds_time[ERA5_VARS].sel(
            time=slice(
                storm["start_time"],
                storm["end_time"]
            ),
            latitude=slice(
                lat_min,
                lat_max
            ),
            longitude=slice(
                lon_min,
                lon_max
            )
        ).compute()

        # ----------------------------------------------------
        # Extract each IBTrACS observation locally
        # ----------------------------------------------------

        for _, obs in storm_obs.iterrows():

            try:

                point = storm_era5.sel(
                    time=obs["ERA5_TIME"],
                    latitude=obs["LAT"],
                    longitude=obs["LON"],
                    method="nearest"
                )

                all_results.append({
                    "SID": obs["SID"],
                    "SEASON": obs["SEASON"],
                    "NAME": obs["NAME"],
                    "ISO_TIME": obs["ISO_TIME"],
                    "LAT": obs["LAT"],
                    "LON": obs["LON"],
                    "USA_WIND": obs["USA_WIND"],
                    "USA_PRES": obs["USA_PRES"],

                    "ERA5_TIME": pd.Timestamp(
                        point["time"].values
                    ),

                    "ERA5_LAT": float(
                        point["latitude"].values
                    ),

                    "ERA5_LON": float(
                        point["longitude"].values
                    ),

                    "u10": float(
                        point["u10"].values
                    ),

                    "v10": float(
                        point["v10"].values
                    ),

                    "t2m": float(
                        point["t2m"].values
                    ),

                    "d2m": float(
                        point["d2m"].values
                    ),

                    "msl": float(
                        point["msl"].values
                    ),

                    "sst": float(
                        point["sst"].values
                    )
                })

            except Exception as point_error:

                print(
                    f"  Point error for {sid}: "
                    f"{point_error}"
                )

        # ----------------------------------------------------
        # Progress
        # ----------------------------------------------------

        print(
            f"[{storm_num:3d}/{total_storms}] "
            f"{storm['NAME']:<12} "
            f"| observations: {len(storm_obs):3d} "
            f"| extracted: {len(all_results):5d}"
        )

    except Exception as storm_error:

        print(
            f"ERROR in storm {sid}: "
            f"{storm_error}"
        )

print("=" * 60)
print("EXTRACTION COMPLETE!")
print("Total extracted observations:", len(all_results))
```

```python
# ============================================================
# BATCH EXTRACTION — BATCH 1
# ============================================================

import gc
import time

BATCH_NUMBER = 1
BATCH_SIZE = 25

start_idx = (BATCH_NUMBER - 1) * BATCH_SIZE
end_idx = min(BATCH_NUMBER * BATCH_SIZE, len(storm_ranges))

batch_storms = storm_ranges.iloc[start_idx:end_idx]

print(f"Starting Batch {BATCH_NUMBER}")
print(f"Cyclones: {start_idx + 1} → {end_idx}")
print("=" * 60)

batch_results = []

ERA5_VARS = ["u10", "v10", "t2m", "d2m", "msl", "sst"]
PADDING = 0.5

batch_start_time = time.time()

for storm_num, (_, storm) in enumerate(batch_storms.iterrows(), start=start_idx + 1):

    sid = storm["SID"]

    # Get IBTrACS observations for this cyclone
    storm_obs = era5_lookup[
        era5_lookup["SID"] == sid
    ].copy()

    if len(storm_obs) == 0:
        continue

    # Small padded geographic region
    lat_min = max(0, storm["min_lat"] - PADDING)
    lat_max = min(30, storm["max_lat"] + PADDING)

    lon_min = max(40, storm["min_lon"] - PADDING)
    lon_max = min(100, storm["max_lon"] + PADDING)

    try:

        # Retrieve only this cyclone's ERA5 block
        storm_era5 = ds_time[ERA5_VARS].sel(
            time=slice(
                storm["start_time"],
                storm["end_time"]
            ),
            latitude=slice(
                lat_min,
                lat_max
            ),
            longitude=slice(
                lon_min,
                lon_max
            )
        ).compute()

        # Match each IBTrACS observation locally
        for _, obs in storm_obs.iterrows():

            point = storm_era5.sel(
                time=obs["ERA5_TIME"],
                latitude=obs["LAT"],
                longitude=obs["LON"],
                method="nearest"
            )

            batch_results.append({
                "SID": obs["SID"],
                "SEASON": obs["SEASON"],
                "NAME": obs["NAME"],
                "ISO_TIME": obs["ISO_TIME"],
                "LAT": obs["LAT"],
                "LON": obs["LON"],
                "USA_WIND": obs["USA_WIND"],
                "USA_PRES": obs["USA_PRES"],

                "ERA5_TIME": pd.Timestamp(
                    point["time"].values
                ),

                "ERA5_LAT": float(
                    point["latitude"].values
                ),

                "ERA5_LON": float(
                    point["longitude"].values
                ),

                "u10": float(point["u10"].values),
                "v10": float(point["v10"].values),
                "t2m": float(point["t2m"].values),
                "d2m": float(point["d2m"].values),
                "msl": float(point["msl"].values),
                "sst": float(point["sst"].values)
            })

        print(
            f"[{storm_num:3d}/264] "
            f"{storm['NAME']:<12} | "
            f"Obs: {len(storm_obs):3d} | "
            f"Batch extracted: {len(batch_results):4d}"
        )

        # Free the cyclone block before moving on
        del storm_era5
        gc.collect()

    except Exception as e:

        print(
            f"ERROR — cyclone {sid} "
            f"({storm['NAME']}): {e}"
        )

# ============================================================
# SAVE BATCH
# ============================================================

batch_df = pd.DataFrame(batch_results)

batch_filename = f"era5_batch_{BATCH_NUMBER:02d}.csv"

batch_df.to_csv(
    batch_filename,
    index=False
)

elapsed = time.time() - batch_start_time

print("=" * 60)
print(f"BATCH {BATCH_NUMBER} COMPLETE")
print(f"Rows extracted: {len(batch_df)}")
print(f"Saved as: {batch_filename}")
print(f"Time taken: {elapsed / 60:.2f} minutes")
```
