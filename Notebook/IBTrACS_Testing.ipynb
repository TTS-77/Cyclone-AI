```python
from google.colab import files

print("Select your ERA5 CSV file:")
era5_upload = files.upload()

print("\nUploaded:")
for filename in era5_upload:
    print(filename)
```

```python
# Import Pandas again
import pandas as pd

# Load the ERA5 dataset
era5_filename = list(era5_upload.keys())[0]

era5 = pd.read_csv(era5_filename)

print("ERA5 dataset loaded successfully!")
print("Shape:", era5.shape)

print("\nColumns:")
print(era5.columns.tolist())

print("\nFirst 5 rows:")
display(era5.head())
```

```python
# Clean ERA5 data

# Convert timestamp to datetime
era5["valid_time"] = pd.to_datetime(era5["valid_time"])

# Convert coordinates and weather variables to numeric
numeric_columns = [
    "latitude",
    "longitude",
    "u10",
    "v10",
    "d2m",
    "t2m",
    "msl",
    "sst"
]

for col in numeric_columns:
    era5[col] = pd.to_numeric(era5[col], errors="coerce")

print("ERA5 data types:")
print(era5.dtypes)

print("\nMissing values:")
print(era5.isna().sum())

print("\nERA5 time range:")
print(era5["valid_time"].min(), "to", era5["valid_time"].max())

print("\nERA5 coordinates:")
print("Latitude:", era5["latitude"].unique())
print("Longitude:", era5["longitude"].unique())
```

```python
import pandas as pd

# Load IBTrACS
ibtracs = pd.read_excel("amphan_ibtracs.csv.xlsx")

# Clean column names
ibtracs.columns = ibtracs.columns.str.strip()

# Convert time
ibtracs["ISO_TIME"] = pd.to_datetime(ibtracs["ISO_TIME"])

# Convert wind and pressure to numeric
ibtracs["USA_WIND"] = pd.to_numeric(ibtracs["USA_WIND"], errors="coerce")
ibtracs["USA_PRES"] = pd.to_numeric(ibtracs["USA_PRES"], errors="coerce")

print("IBTrACS loaded successfully!")
print("Shape:", ibtracs.shape)

display(ibtracs.head())
```

```python
# Check how much of the IBTrACS track falls inside our ERA5 dataset

print("IBTrACS time range:")
print(ibtracs["ISO_TIME"].min(), "to", ibtracs["ISO_TIME"].max())

print("\nERA5 time range:")
print(era5["valid_time"].min(), "to", era5["valid_time"].max())

print("\nIBTrACS coordinate range:")
print(
    "Latitude:", ibtracs["LAT"].min(), "to", ibtracs["LAT"].max()
)
print(
    "Longitude:", ibtracs["LON"].min(), "to", ibtracs["LON"].max()
)

print("\nERA5 coordinate range:")
print(
    "Latitude:", era5["latitude"].min(), "to", era5["latitude"].max()
)
print(
    "Longitude:", era5["longitude"].min(), "to", era5["longitude"].max()
)

# Find IBTrACS observations whose timestamp exists in ERA5
time_overlap = ibtracs[
    ibtracs["ISO_TIME"].isin(era5["valid_time"])
]

print("\nIBTrACS observations with matching ERA5 timestamps:",
      len(time_overlap))

display(time_overlap[
    ["ISO_TIME", "LAT", "LON", "USA_WIND", "USA_PRES"]
].head(15))
```
